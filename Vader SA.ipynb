```python
import pandas as pd
```

# Constrituencies

```python
data=pd.read_csv(r'csvs\Cleaned_constituency_data.csv')
```

```python
df=data.copy()
```

```python
df
```

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

# Initialize the sentiment analyzer
analyzer = SentimentIntensityAnalyzer()

# Apply sentiment analysis
df['vader_scores'] = df['Extracted_Text'].apply(lambda x: analyzer.polarity_scores(str(x)))
df['vader_compound'] = df['vader_scores'].apply(lambda score_dict: score_dict['compound'])
df['vader_pos'] = df['vader_scores'].apply(lambda score_dict: score_dict['pos'])
df['vader_neu'] = df['vader_scores'].apply(lambda score_dict: score_dict['neu'])
df['vader_neg'] = df['vader_scores'].apply(lambda score_dict: score_dict['neg'])
```

```python
sentiment_over_time = df.groupby('Year')['vader_compound'].mean().reset_index()
```

```python
import plotly.express as px

fig = px.line(sentiment_over_time, x='Year', y='vader_compound',
              title='Average Sentiment Over Time (VADER)',
              labels={'vader_compound': 'Compound Sentiment Score'})
fig.show()
```

```python
df = df[df['Constituency'] != "OBS"]
```

```python
fig = px.box(df, x='Region/Authority', y='vader_compound',
             title='Sentiment Distribution by Authority',
             labels={'vader_compound': 'Compound Sentiment Score'})
fig.update_layout(xaxis_tickangle=-45)
fig.show()
```
